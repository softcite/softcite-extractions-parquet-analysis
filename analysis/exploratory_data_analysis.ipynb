```python
import pandas as pd

# Create mentions dataframe
mentions_file = '../data/softcite-extractions-oa-data/p01_one_percent_random_subset/mentions.pdf.parquet'
mentions_df = pd.read_parquet(mentions_file, engine='fastparquet')
mentions_df.head()
```

```python
# Create papers dataframe
papers_file = '../data/softcite-extractions-oa-data/p01_one_percent_random_subset/papers.parquet'
papers_df = pd.read_parquet(papers_file, engine='fastparquet')
papers_df.head()
```

```python
import matplotlib.pyplot as plt

# Get count of years
year_counts = papers_df['published_year'].value_counts().sort_index()

# Create barchart representing the distribution of software mentions detected by year
plt.figure(figsize=(20, 15))
year_counts.plot(kind ='bar')

plt.title('Count of Software Mentions in Academic Papers by Year')
plt.xlabel('Year')
plt.ylabel('Count')

plt.show()
```
